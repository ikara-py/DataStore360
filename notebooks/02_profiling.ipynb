**Check for missing Values**

```python
m_names = len(df) - df['Customer Name'].count()
m_Postal_codes = len(df) - df['Postal Code'].count()
m_sales = len(df) - df['Sales'].count()
m_quantity = len(df) - df['Quantity'].count()
m_ship_mode = len(df) - df['Ship Mode'].count()
m_ship_date = len(df) - df['Ship Date'].count()

print(f"[+] missing Customer Name : {m_names}\n[+] missing Postal Codes : {m_Postal_codes}\n[+] missing Sales : {m_sales}\n[+] missing Quantity : {m_quantity}\n[+] missing Ship Mode : {m_ship_mode}\n[+] missing Ship Date : {m_ship_date}")
```

**Sales**

```python
s_mean = df['Sales'].mean()
s_min = df['Sales'].min()
s_max = df['Sales'].max()
s_median = df['Sales'].median()

print(f"[+] Sales Mean : {s_mean}\n"f"[+] Sales Min : {s_min}\n"f"[+] Sales Max : {s_max}\n"f"[+] Sales Median : {s_median}")
```

**Quantity**

```python
q_mean = df['Quantity'].mean()
q_min = df['Quantity'].min()
q_max = df['Quantity'].max()
q_median = df['Quantity'].median()


print(f"[+] Quantity Mean : {q_mean}\n[+] Quantity Min : {q_min}\n[+] Quantity Max : {q_max}\n[+] Quantity Median : {q_median}")
```

**Discount**

```python
d_mean = df['Discount'].mean()
d_min = df['Discount'].min()
d_max = df['Discount'].max()
d_median = df['Discount'].median()

print(f"[+] Discount Mean : {d_mean}\n[+] Discount Min : {d_min}\n[+] Discount Max : {d_max}\n[+] Discount Median : {d_median}")
```
